## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Task Type: Binary Classification

Justification: We are framing the problem as predicting whether a user will convert (1) or not convert (0) within a specific window. This fits a binary classification formulation because the outcome is discrete, mutually exclusive, and serves directly to route high-intent leads to sales support while reducing manual intervention for low-intent users.

```python
import pandas as pd
import numpy as np

# Create a synthetic dataset in memory
np.random.seed(42)
n_samples = 200

df = pd.DataFrame({
    "feature_1": np.random.normal(loc=10, scale=2, size=n_samples),
    "feature_2": np.random.exponential(scale=5, size=n_samples),
    "target": np.random.choice([0, 1], size=n_samples, p=[0.88, 0.12])
})

# Check distribution of target
print("Class Balance:")
print(df["target"].value_counts(normalize=True))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Primary Metric: PR-AUC (Precision-Recall Area Under the Curve)

Justification: Given class imbalance (conversions are typically rare), ROC-AUC can present an overly optimistic view. PR-AUC explicitly measures performance on the positive class without being inflated by true negatives.

Baseline Target: A PR-AUC threshold of > 0.45 (compared to a baseline random guess PR-AUC equal to the positive class prevalence of ~0.12).

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from sklearn.metrics import average_precision_score

# Simulated dummy baseline check
y_true = df["target"]
y_dummy_pred = np.full_like(y_true, fill_value=y_true.mean(), dtype=float)

baseline_prauc = average_precision_score(y_true, y_dummy_pred)
print(f"Random/Naive Baseline PR-AUC: {baseline_prauc:.4f}")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why ML Beats Fixed Rules: Static if/else rules struggle with non-linear feature interactions (e.g., high session count combined with low time-on-site vs. low session count with high intent actions). Machine learning dynamically adapts to high-dimensional interactions and shifting behavioral distributions without requiring manual updates to hardcoded rule thresholds.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Demonstrate complex feature correlations/variance
feature_cols = [c for c in df.columns if c != "target"]
print("Feature Variance Check:")
print(df[feature_cols].var())

print("\nFeature Correlations with Target:")
print(df.corr(numeric_only=True)["target"].sort_values(ascending=False))
```
